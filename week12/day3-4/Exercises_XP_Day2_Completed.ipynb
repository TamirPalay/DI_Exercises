## 🌟 Exercise 1: Traditional vs. Modern NLP: Comparative Analysis
1) Complete the table:

| Aspect | Traditional NLP | Modern NLP |
|---|---|---|
| Feature Engineering | Manual — linguists and engineers hand-craft features (POS tags, n-grams, TF-IDF, parse trees) for each task | Automatic — deep networks learn features directly from raw text during training |
| Word Representations | Static — one fixed vector per word regardless of context (e.g., Bag-of-Words, TF-IDF, Word2Vec, GloVe) | Contextual — a word's vector changes depending on the surrounding sentence (e.g., BERT, GPT embeddings) |
| Model Architectures | Shallow — linear/statistical models (Naïve Bayes, HMMs, CRFs, shallow feedforward nets) | Deep — many-layered neural networks, mainly Transformers with self-attention |
| Training Methodology | Task-specific — a model is trained from scratch for one narrow task using a labeled dataset | Pre-training + fine-tuning — a model first learns general language patterns on massive unlabeled text, then is adapted to specific tasks with comparatively little labeled data |
| Key Examples | Naïve Bayes, Hidden Markov Models, Conditional Random Fields, Word2Vec, GloVe | BERT, GPT, T5, RoBERTa |
| Advantages | Interpretable, computationally cheap, works with small datasets, easy to debug | Captures context and nuance, transfers across tasks, state-of-the-art accuracy, less manual engineering |
| Disadvantages | Can't capture context or long-range dependencies, brittle to unseen phrasing, requires heavy manual feature work | Expensive to train/host, data- and compute-hungry, less interpretable, can inherit biases from training data |

2) Discuss: How did the shift to modern NLP impact scalability and efficiency?

Moving from hand-engineered features to learned representations meant a single pre-trained model could be reused across many downstream tasks instead of building a bespoke pipeline for each one, which dramatically cut development time and cost. Transformer architectures replaced the sequential processing of RNNs with fully parallelizable self-attention, letting training take full advantage of modern GPU/TPU hardware and scale to billions of parameters and web-scale corpora. Transfer learning meant that most of the expensive learning (pre-training) only had to happen once, after which fine-tuning on a specific task needed only a small labeled dataset and a fraction of the compute. This made it economically feasible for far more organizations to build high-quality NLP systems, since they could start from a strong pre-trained model rather than collecting large labeled datasets themselves. At the same time, contextual representations meant one model could handle ambiguity, sarcasm, and long-range dependencies that static, feature-engineered pipelines struggled with, improving quality alongside scalability. The trade-off is that the biggest modern models are themselves very expensive to train and serve, so efficiency gains at the task level are partly offset by the scale of the underlying models. Techniques like distillation, quantization, and parameter-efficient fine-tuning (e.g., LoRA) have emerged specifically to recover some of that efficiency while keeping the benefits of pre-training.


## 🌟 Exercise 2: LLM Architecture and Application Scenarios
For each, describe (a) core architectural differences, (b) a real application, (c) why it fits.

### BERT
- **Architecture:** Encoder-only, bidirectional — every token attends to both the tokens before and after it. Pre-trained with Masked Language Modeling (predict randomly masked tokens using full context) and, originally, Next Sentence Prediction.
- **Application:** Named Entity Recognition (NER) and document classification, e.g., extracting company names and flagging sentiment in financial news articles.
- **Why it fits:** NER and classification require understanding a token's full surrounding context (both left and right) to disambiguate meaning — e.g., knowing whether "Amazon" refers to the company or the river depends on words that come after it as much as before. BERT's bidirectional attention captures exactly that, and since these tasks only need a rich understanding of existing text (not new text generation), an encoder-only model is sufficient and efficient.

### GPT
- **Architecture:** Decoder-only, unidirectional (autoregressive) — each token can only attend to previous tokens, and the model is trained with Causal Language Modeling (predict the next token given everything before it).
- **Application:** Open-ended conversational assistants and creative writing tools, e.g., a chatbot that drafts marketing copy or holds a multi-turn conversation.
- **Why it fits:** Generation is inherently sequential — you produce one token, then the next, conditioned only on what's already been written. The causal, left-to-right structure of GPT mirrors this generation process directly, making it naturally suited to producing fluent, open-ended text rather than just analyzing existing text.

### T5
- **Architecture:** Encoder-decoder — an encoder builds a bidirectional representation of the input, and a decoder generates output tokens autoregressively while attending back to that representation (cross-attention). T5 frames every task as "text-to-text."
- **Application:** Machine translation and abstractive summarization, e.g., translating English support tickets into Spanish or condensing long reports into short summaries.
- **Why it fits:** These tasks need to fully understand a source text (favoring bidirectional encoding) and then produce a different, freely generated text in response (favoring autoregressive decoding). Only an encoder-decoder architecture gives you both halves in one model — the encoder can look at the entire source sentence to build a faithful representation, while the decoder generates a fluent, appropriately ordered target sentence conditioned on it.


## 🌟 Exercise 3: Benefits and Ethics of Pre-training
- Benefits (in my own words):
  1) **Improved generalization:** By training on enormous, varied text, the model learns broad language patterns rather than quirks of a single small dataset, so it performs better on new, unseen data.
  2) **Less labeled data:** Since most of the "learning to understand language" work already happened during pre-training, fine-tuning for a new task only needs a small labeled dataset instead of a huge one built from scratch.
  3) **Faster fine-tuning:** Starting from weights that already encode grammar, world knowledge, and semantics means fine-tuning converges in far fewer training steps than training a model from random initialization.
  4) **Transfer learning:** Knowledge learned on one broad task (predicting text) transfers to many different downstream tasks — sentiment analysis, translation, Q&A — without re-learning language from zero each time.
  5) **Robustness:** Exposure to diverse text (different topics, styles, and writing conventions) during pre-training makes the model less brittle to typos, unusual phrasing, or domain shifts than a model trained on one narrow dataset.

- **Ethical concerns:** Pre-training on massive, largely unfiltered web text means models can absorb and reproduce societal biases (racial, gender, cultural stereotypes) present in that data. They can also generate convincing misinformation or hallucinated "facts" that sound authoritative. Because base models are general-purpose, they can be misused for spam, phishing, disinformation campaigns, or generating harmful content. Training data can also include private or copyrighted material scraped without clear consent, raising privacy and IP concerns.

- **Mitigation strategies:** Curating and filtering training data to reduce toxic, biased, or low-quality content; running bias audits and fairness evaluations before deployment; applying RLHF (reinforcement learning from human feedback) or similar alignment techniques to steer outputs toward safe, helpful behavior; adding safety classifiers/filters at inference time to catch harmful outputs; and being transparent with model cards and documentation about known limitations so downstream users understand the risks.


## 🌟 Exercise 4: Transformer Architecture Deep Dive
### Self-Attention & Multi-Head Attention
**How self-attention works:** Each input token embedding is projected into three vectors — Query (Q), Key (K), and Value (V) — using learned weight matrices. For a given token, its Query vector is compared (dot product) against every token's Key vector in the sequence, producing a raw similarity score for each pair. These scores are scaled (divided by the square root of the key dimension) and passed through a softmax, turning them into attention weights that sum to 1 — essentially "how much should this token pay attention to every other token." Each token's output is then a weighted sum of all tokens' Value vectors, using those attention weights. The result is that every token's new representation is a blend of the whole sentence's information, weighted by relevance.

**Why multi-head attention helps:** A single attention head can only learn one type of relationship pattern between tokens at a time (it computes one set of attention weights). Multi-head attention runs several attention operations in parallel, each with its own learned Q/K/V projections, so different heads can specialize — one might track subject-verb agreement, another might track coreference (which noun a pronoun refers to), another might track adjective-noun relationships. Their outputs are concatenated and linearly projected back to the model dimension, giving a richer, multi-faceted representation than a single head could produce alone.

**Example sentence:** "The trophy didn't fit in the suitcase because it was too small."
- One head might focus on resolving "it" → "suitcase" (coreference resolution) versus "it" → "trophy," using surrounding context ("too small" fitting a container more than a trophy).
- Another head might focus on the syntactic dependency between "didn't fit" and "because," linking the cause-and-effect structure of the sentence.
- A third head might attend locally, linking "suitcase" to its article "the" and to "in," tracking simple grammatical structure.

### Pre-training Objectives
**MLM vs CLM:** Masked Language Modeling (used by BERT) randomly masks a percentage of tokens in the input and trains the model to predict them using both left and right context — this requires bidirectional attention. Causal Language Modeling (used by GPT) trains the model to predict the next token using only the preceding tokens — this requires unidirectional (causal) attention so the model can't "cheat" by seeing the future.

**When to prefer each:** MLM is better suited to tasks that need a deep, holistic understanding of existing text — classification, extraction, question answering — since bidirectional context improves comprehension. CLM is better suited to tasks that involve generating new text left-to-right, such as open-ended dialogue, story writing, or code completion, since the model is trained in exactly the way it will be used at inference time.

**Next Sentence Prediction (NSP):** Early BERT used NSP (predicting whether one sentence follows another) because the researchers believed it would help the model learn relationships between sentences, which is useful for tasks like question answering. However, later research (e.g., RoBERTa) found NSP added little benefit and sometimes hurt performance, partly because the task was too easy (the negative examples were often from entirely different documents, making the signal trivial). Modern models tend to drop NSP in favor of just training longer on MLM with better data, which was found to be simpler and more effective.

### Transformer Model Selection
- **Sentiment on reviews (positive/negative/neutral):** Encoder-only. This is a classification task over existing text — the model needs a rich bidirectional understanding of the whole review, not text generation. Encoder-only models like BERT are efficient and well-suited here since only a single label output is required.
- **Conversational chatbot (creative responses):** Decoder-only. Generating open-ended, creative conversational text is inherently sequential and autoregressive; the model needs to produce novel word sequences one token at a time, which is exactly what decoder-only models like GPT are trained to do.
- **Technical document translation (EN→ES):** Encoder-decoder. Translation needs a full bidirectional understanding of the source sentence (encoder) combined with fluent, correctly-ordered generation in the target language (decoder), with cross-attention letting the decoder reference the source at each generation step — exactly what models like T5 or the original Transformer were designed for.

### Positional Encoding
**Purpose:** Self-attention itself has no built-in notion of word order — it treats the input as an unordered set of tokens, since the same attention computation would occur regardless of token position. Positional encodings inject information about each token's position in the sequence (e.g., via sinusoidal functions or learned position embeddings) so the model can distinguish "dog bites man" from "man bites dog."

**Example failure without positional encoding:** Without positional information, the sentences "The chef served the customer" and "The customer served the chef" would produce identical sets of token representations from self-attention (same words, same pairwise attention scores regardless of order), so the model couldn't tell who is serving whom — a critical distinction the sentence is trying to convey.


## 🌟 Exercise 5: BERT Variations: Choose Your Detective
- **Scenario 1 (mobile, limited resources):** **DistilBERT.** It's trained via knowledge distillation to be ~40% smaller and ~60% faster than BERT while retaining ~97% of its language understanding, making it the best fit when memory and latency are tightly constrained, as on a mobile device.
- **Scenario 2 (legal docs, high accuracy):** **RoBERTa.** It removes NSP, trains on much more data with larger batches and longer sequences, and generally outperforms BERT on accuracy-sensitive benchmarks, which matters most when precision on complex legal text is the priority over efficiency.
- **Scenario 3 (multilingual support):** **XLM-RoBERTa.** It's pre-trained on a massive multilingual corpus spanning 100 languages with a shared vocabulary, letting one model handle customer support across languages instead of needing separate models per language.
- **Scenario 4 (efficient pretraining, token replacement detection):** **ELECTRA.** Its defining innovation is replacing MLM with "replaced token detection" — a generator swaps some tokens and a discriminator learns to spot which tokens were replaced — which trains on every token (not just ~15% masked ones), making pre-training far more compute-efficient.
- **Scenario 5 (efficient NLP, constrained environments):** **ALBERT.** It uses factorized embedding parameterization and cross-layer parameter sharing to drastically cut parameter count versus BERT, making it memory-efficient while retaining strong performance — well suited to constrained environments.

Comparison table:

| Model | Training differences | Size/Efficiency | Innovations | Ideal use cases |
|---|---|---|---|---|
| RoBERTa | Removes NSP, trains longer with larger batches, more data, dynamic masking | Similar size to BERT, no efficiency gain — focused purely on accuracy | Dynamic masking (masks change each epoch instead of being fixed), more/better training data | High-accuracy tasks where compute budget is less of a concern, e.g., research and analysis on complex documents |
| ALBERT | Same MLM objective but shares parameters across layers | Much smaller parameter count than BERT for similar depth | Factorized embedding parameterization, cross-layer parameter sharing, sentence-order prediction (instead of NSP) | Memory-constrained deployments needing BERT-level depth without BERT-level parameter count |
| DistilBERT | Trained via knowledge distillation from a full BERT "teacher" model | ~40% smaller, ~60% faster inference, retains ~97% of BERT's performance | Distillation loss combining teacher's soft predictions with standard MLM loss | Real-time or on-device applications like mobile apps and latency-sensitive services |
| ELECTRA | Uses a generator-discriminator setup; discriminator predicts replaced vs. original tokens on every position | More sample- and compute-efficient pre-training than MLM-based models at equivalent model size | Replaced Token Detection (RTD) objective instead of masking | Settings where pre-training compute budget is limited but strong downstream performance is still needed |
| XLM-R | Pre-trained on a large multilingual corpus (CommonCrawl) covering 100 languages with a shared subword vocabulary | Larger vocabulary/embedding table than monolingual BERT, but one model replaces many | Cross-lingual pre-training without needing parallel/translated sentence pairs | Multilingual applications like global customer support, cross-lingual search, and translation-adjacent tasks |


## 🌟 Exercise 6: Softmax Temperature: The Randomness Regulator
### 1) Temperature Scenarios
- **T = 0.2 (low):** The probability distribution over the next token becomes much sharper/peakier — the model almost always picks the single most likely token. Output becomes very deterministic, repetitive, and "safe," but can feel dull or get stuck in loops.
- **T = 1.5 (high):** The distribution flattens, giving lower-probability tokens a much better chance of being selected. Output becomes more diverse and creative, but also more likely to be incoherent, off-topic, or nonsensical, since the model is willing to pick surprising, less-likely words.
- **T = 1.0 (neutral):** This applies the raw softmax probabilities exactly as computed by the model with no rescaling, giving a balanced mix of likely and occasionally surprising word choices — the model's "default" calibration.

### 2) Application Design
- **Bedtime stories for children:** I'd use a moderate-to-slightly-elevated temperature (roughly 0.7–1.0), possibly combined with top-p (nucleus) sampling, to keep the story imaginative and varied — introducing fun, unexpected plot turns and character details — while still staying grammatically coherent and easy to follow for a child audience. Too low and every story would feel formulaic; too high and the plot could become confusing or drift into inappropriate/strange territory.
- **Financial report summaries:** I'd use a low temperature (roughly 0.1–0.3), or even greedy/deterministic decoding, since accuracy and consistency matter far more than creativity here. A low temperature reduces the chance the model invents a plausible-sounding but incorrect number or claim, keeping the summary faithful to the source report.

### 3) Temperature & Bias
Temperature doesn't create bias — it changes how much of the model's underlying (already-biased) probability distribution gets exposed in the output. At low temperature, the model consistently outputs its single most probable completion, which means if that top completion reflects a stereotype learned from training data, it will be produced very reliably and predictably every time. At high temperature, the model samples further into the tail of the distribution, which can sometimes surface less common (and possibly less biased) completions, but can just as easily surface other biased or nonsensical text that happens to have some non-trivial probability. **Example:** If a model is prompted with "The nurse walked into the room, and she..." and its training data strongly associates "nurse" with female pronouns, then at low temperature the model will almost always default to the same gendered continuation, consistently reinforcing that stereotype. At high temperature, gendered assumptions might appear less consistently (since other tokens get more of a chance), but the underlying bias in the model's learned probabilities hasn't actually been removed — temperature only reshapes sampling, it doesn't debias the model.


### (Optional) Quick Generation Demo Across Temperatures
Note: Requires `transformers` and model download; skip if offline.


```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "gpt2"
prompt = "The future of language models will"

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name),
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True)
    print("--- Temperature:", T)
    print(out[0]["generated_text"])  # Inspect how style changes with T
```
