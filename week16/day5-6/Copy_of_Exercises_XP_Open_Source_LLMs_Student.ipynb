4) Write a 3?5 sentence comparative paragraph.  
5) Answer the healthcare prompt (1?2 sentences) about retraining on clinical data.

**Deliverables:** paragraph + healthcare answer.

```python

# TODO: capture definitions and key characteristics
open_source_levels = {
    "Fully Open": {
        "definition": "Everything is available",
        "what_is_open": ["weight", "training code", "training data"],
        "what_you_can_do": ["Reproduce the model"],
        "what_you_cannot_do": ["Nothing"],
    },
    "Weights Released": {
        "definition": "Only model parameters released",
        "what_is_open": ["wieghts", "hyperparameters"],
        "what_you_can_do": ["fine tune", "use directly"],
        "what_you_cannot_do": ["retrain"],
    },
    "Architecture Only": {
        "definition": "Blueprint of model",
        "what_is_open": ["strucutral layers and logic"],
        "what_you_can_do": ["understand and retrain model"],
        "what_you_cannot_do": ["Fine tune or use directly"],
    },
}
open_source_levels
```

```python

# TODO: fill the comparison table (keep markdown formatting)
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Weights, training code, and training data | Full reproducibility — you can retrain from scratch, audit the data, and modify any part of the pipeline |
| Weights Released | Only the trained model parameters (and often hyperparameters) | You can run and fine-tune the model, but you can't reproduce or fully understand how it was originally trained |
| Architecture Only | The model's code/blueprint (layers, structure), no trained weights | You can study and reimplement the design, but you must train it yourself from scratch to get a usable model |"""
print(comparison_table)
```

```python

# TODO: 3-5 sentence comparative paragraph
comparative_paragraph = """The three levels of openness represent a spectrum of transparency and control over a language model. Fully Open models give you everything — weights, training code, and training data — so you can reproduce, audit, and retrain the model from the ground up. Weights Released models sit in the middle: you get a ready-to-use model that you can run or fine-tune on your own data, but you have no visibility into the original training process or data. Architecture Only releases give you the design blueprint with no trained weights at all, meaning you understand how the model is built but must invest significant compute and data to train it yourself. In practice, most teams work with Weights Released models since they balance usability with reasonable compute requirements, reserving Fully Open models for cases that demand full auditability, such as regulated industries."""
print(comparative_paragraph)
```

```python

# TODO: 1-2 sentence healthcare prompt answer
healthcare_prompt_answer = """For retraining on sensitive clinical data, a Fully Open model is the safest choice because it lets you audit the original training data and pipeline for biases or compliance gaps, and retrain entirely within your own controlled environment without depending on an external provider. A Weights Released model could still be fine-tuned on clinical data, but you'd have no way to verify what the base model was originally trained on, which is a real risk for healthcare compliance and explainability."""
print(healthcare_prompt_answer)
```
