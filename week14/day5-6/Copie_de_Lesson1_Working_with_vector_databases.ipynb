## Augmentation

**Augmentation** : Creating a single unified prompt out of the original prompt + the retreived chunks

> Let's say we went ahead and created a "cat-index" containing cat facts

```python
# Initialize index client
# index-auto = pc.Index(name=index_name)

# View index stats
# index.describe_index_stats()


def augment(user_prompt):
  ret_chunks = find_similar_questions_auto(user_prompt)

  new_prompt = \
  f"""
  Using only the provided retrieved documents, answer the following question. Do not add any external knowledge.

  ## Original question
  {user_prompt}

  ## Retreived documents
  {ret_chunks}
  """

  return new_prompt


print(augment("Which metropolis has the highest number of people?"))
```

```python

def chat_function(message, history):
    print(f"MESSAGE : {message}")
    # Prepare messages for the API call
    # `history` is a list of [user_message, assistant_message] pairs from Gradio.
    # OpenRouter API expects a list of dictionaries with 'role' and 'content'.
    formatted_messages = []

    # Defensively filter history to ensure each item is a list/tuple of two strings
    cleaned_history = []
    for item in history:
        if isinstance(item, (list, tuple)) and len(item) == 2 and \
           isinstance(item[0], str) and isinstance(item[1], str):
            cleaned_history.append(item)
        else:
            print(f"Skipping malformed history item: {item}") # Log for debugging

    for user_msg, assistant_msg in cleaned_history:
        formatted_messages.append({"role": "user", "content": user_msg})
        if assistant_msg: # only add assistant message if it exists
            formatted_messages.append({"role": "assistant", "content": assistant_msg})
    formatted_messages.append({"role": "user", "content": augment(message)})

    # First API call with reasoning
    response = requests.post(
      url="https://openrouter.ai/api/v1/chat/completions",
      headers={
        "Authorization": f"Bearer {OPENROUTER_API_KEY}",
        "Content-Type": "application/json",
      },
      data=json.dumps({
        "model": "openrouter/free",
        "messages": formatted_messages, # Corrected messages parameter
        "reasoning": {"enabled": True}
      })
    )

    # Extract the assistant message with reasoning_details
    response_json = response.json()
    print(response_json)
    if 'choices' in response_json and len(response_json['choices']) > 0:
        # Ensure the content is a string, even if it's None
        content = response_json['choices'][0]['message'].get('content', '')
        return content if content is not None else ''
    elif 'error' in response_json:
        return f"Error from OpenRouter API: {response_json['error']['message']}"
    else:
        return "An unexpected error occurred or the API response format changed."
```
