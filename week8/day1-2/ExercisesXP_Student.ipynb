Problem Statement
The objective is to predict the probability that a loan applicant will default on their loan — defined as failing to meet repayment obligations within 90 days of the due date — in order to support credit officers in making data-driven approval decisions that minimize financial risk while ensuring fair lending practices.

Data Types
Personal applicant details such as age, employment status, and years at current job are needed because financial stability indicators correlate strongly with repayment reliability. Credit history data — including credit score, number of open accounts, prior defaults, and credit utilization rate — is among the most predictive signals for default risk, as it directly reflects an applicant's past borrowing behavior. Loan-specific attributes such as loan amount, interest rate, loan term, and purpose matter because higher loan-to-income ratios and longer terms are associated with elevated default rates. Repayment history on existing or past loans is critical because missed or late payments are the strongest behavioral indicator of future default. Finally, income and expense data, including monthly income, existing debt obligations, and debt-to-income ratio, quantify the applicant's actual capacity to service new debt.

Data Sources, Risks & Constraints
Internal records from the financial institution — loan applications, repayment logs, and account histories — form the primary dataset and can be accessed directly through the institution's data warehouse or CRM system. Credit bureaus such as Experian, Equifax, or local equivalents provide standardized credit scores and external borrowing history, and would be integrated via licensed API access with applicant consent. Government or open financial datasets can supplement demographic and macroeconomic context such as regional unemployment rates. However, this project carries significant risks: applicant data is highly sensitive and subject to regulations such as GDPR or local financial privacy laws, requiring strict access controls and anonymization protocols. Sampling bias is a serious concern — if historical approvals excluded certain demographics, the model may perpetuate discriminatory lending patterns. Data quality issues such as incomplete repayment records or outdated income figures must be addressed through validation pipelines, and all modeling decisions should be documented to satisfy regulatory audit requirements.

`````python

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, average_precision_score,
    confusion_matrix, classification_report
)

# Replace with real labels and predicted probabilities from your trained model
y_true = [0, 1, 0, 1, 0, 0, 1, 0, 1, 0]
y_pred_proba = [0.05, 0.80, 0.10, 0.65, 0.20, 0.15, 0.70, 0.30, 0.85, 0.25]

# Lower threshold (e.g. 0.3) to prioritize recall on the default class
# In production, tune this on the validation set using the PR curve
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy      :", round(accuracy_score(y_true, y_pred), 4))
print("Precision     :", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall        :", round(recall_score(y_true, y_pred, zero_division=0), 4))  # Priority metric
print("F1-score      :", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC       :", round(roc_auc_score(y_true, y_pred_proba), 4))            # Overall discrimination
print("PR-AUC        :", round(average_precision_score(y_true, y_pred_proba), 4)) # Better under imbalance

print("\nConfusion Matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification Report:\n", classification_report(y_true, y_pred, zero_division=0))
`````
