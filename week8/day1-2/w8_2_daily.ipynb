```python
## What is Machine Learning?

ML is the practice of training algorithms to learn patterns from data and make predictions
or decisions without being explicitly programmed for each task.

It is essential for data analysts because it scales pattern recognition beyond what manual
analysis can handle, enabling predictions on large, complex datasets.

---

## Industry Applications

**Finance** — Banks use ML to detect fraudulent transactions in real time.

**Healthcare** — Hospitals use ML to predict patient readmission risk from medical records.

**Retail** — E-commerce platforms use ML to power personalised product recommendation engines.

---

## Types of Machine Learning

### Supervised Learning
The model learns from labeled data — each input has a known output.

*Example: Predicting whether a loan applicant will default, using historical approved loans.*

### Unsupervised Learning
The model finds hidden structure in unlabeled data with no predefined output.

*Example: Grouping customers into segments based on purchasing behaviour.*

### Reinforcement Learning
An agent learns by interacting with an environment and receiving rewards or penalties.

*Example: Training a robot to navigate a maze by rewarding progress toward the exit.*

---

## The ML Model Process

### 1. Feature Selection
Identify the input variables most relevant to the prediction target,
removing noise and redundant columns that could mislead the model.

### 2. Model Selection
Choose the algorithm that best fits the problem type, data size, and interpretability
requirements — for example, Logistic Regression for simple binary classification
or XGBoost for complex tabular data.

### 3. Model Evaluation
Measure performance on unseen data using appropriate metrics such as Accuracy, F1-score,
ROC-AUC for classification, or RMSE for regression, and iterate until results are satisfactory.
```
